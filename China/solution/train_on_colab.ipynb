**First:** Runtime → Change runtime type → **T4 GPU** → Save.

**Before running:** on your computer, zip your `yolo_data` folder into `yolo_data.zip` and upload it and `train.py` to a folder `glyph_project` in Google Drive (My Drive).

```
!nvidia-smi
!pip -q install ultralytics
```

```
from google.colab import drive
drive.mount('/content/drive')
PROJ = '/content/drive/MyDrive/glyph_project'
```

Copy the data to the Colab disk (reading thousands of images straight from Drive is very slow).

```
!cp "$PROJ/train.py" /content/
!unzip -q -o "$PROJ/yolo_data.zip" -d /content/
!ls /content/yolo_data
```

Point `data.yaml` at the new location (your local copy probably has a Windows/Mac path in it).

```
import yaml
p = '/content/yolo_data/data.yaml'
cfg = yaml.safe_load(open(p))
cfg['path'] = '/content/yolo_data'
yaml.safe_dump(cfg, open(p, 'w'), sort_keys=False)
print(open(p).read())
```

Train. Results and weights are written to Drive, so they survive if Colab disconnects.

```
%cd /content
!python train.py --data /content/yolo_data/data.yaml --epochs 40 --project "$PROJ/runs"
```

**If Colab disconnected mid-training:** rerun cells 1–4, then run this instead of the training cell.

```
%cd /content
!python train.py --resume --project "$PROJ/runs"
```

Your trained model is at `My Drive/glyph_project/runs/glyphs/weights/best.pt`. Download it to use locally:
